## **Data Preprocessing and splitting**


```python
def preprocess_text(text):
  text = text.lower()
  text = re.sub(r'\d+', '', text)
  text = re.sub(r'[^\w\s]', '', text)
  stop_words = set(stopwords.words('english'))
  text= ' '.join(word for word in text.split() if word not in stop_words)
  lemmatizer = WordNetLemmatizer()
  text = ' '.join(lemmatizer.lemmatize(word) for word in text.split())
  return text
```

```python
train_data['DESCRIPTION'] = train_data['DESCRIPTION'].apply(preprocess_text)
test_data['DESCRIPTION'] = test_data['DESCRIPTION'].apply(preprocess_text)
```

```python
tfidf_vectorizer = TfidfVectorizer(stop_words="english",max_features=100000)
```

```python
X_train_tfidf = tfidf_vectorizer.fit_transform(train_data['DESCRIPTION'])
X_test_tfidf = tfidf_vectorizer.transform(test_data['DESCRIPTION'])
```

```python
#Encode the targert lables
encider = LabelEncoder()
y_train = encider.fit_transform(train_data['GENRE'])
```

```python
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(X_train_tfidf ,y_train, test_size=0.2, random_state=42)
```

## **Naives Bayes Model**

```python
nb_model = MultinomialNB()
nb_model.fit(X_train_split, y_train_split)
```

```python
y_val_pred_nb = nb_model.predict(X_val_split)
print('Navie Bayes Model')
print(classification_report(y_val_split, y_val_pred_nb,target_names=encider.classes_))
print(f'Accuracy: {accuracy_score(y_val_split, y_val_pred_nb):.4f}\n')
```

## **Logistic Regression Model**

```python
lr_model = LogisticRegression()
lr_model.fit(X_train_split, y_train_split)
```

```python
y_val_pred_lr = lr_model.predict(X_val_split)
print('Logistic Regression Model Evaluation')
print(classification_report(y_val_split,y_val_pred_lr,target_names=encider.classes_))
print(f'Accuracy: {accuracy_score(y_val_split, y_val_pred_lr):.4f}\n')
```

## **SVC Model**

```python
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(X_train_tfidf ,y_train, test_size=0.2, random_state=42)
```

```python
from sklearn.svm import LinearSVC
svc_model = LinearSVC()
svc_model.fit(X_train_split, y_train_split)
```

```python
y_val_pred_svc = svc_model.predict(X_val_split)
print('SVC Model Evaluation:')
print(classification_report(y_val_split,y_val_pred_svc,target_names=encider.classes_))
print(f'Accuracy: {accuracy_score(y_val_split, y_val_pred_svc):.4f}\n')
```

# **Make Predictions on the test data using the SVM  Model**

```python
y_test_pred = svc_model.predict(X_test_tfidf)

#Decode the predicted labels to genre names
y_test_pred_genres = encider.inverse_transform(y_test_pred)

#Add the predicted genres to the test data
test_data['PREDICTED_GENRE'] = y_test_pred_genres

#For the ID of 3,4 and 5
ID = [3,4,5]

filtered_test_data = test_data[test_data['ID'].isin(ID)]

#Display the 3,4and 5th row of the test data with predictions
print(filtered_test_data [['ID','TITLE','PREDICTED_GENRE']].head())
```

## **Predict the genre of the new discription**

```python
new_des = ("The Shawshank Redemption tells the inspiring story of Andy Dufresne, a banker sentenced to life in a brutal prison who maintains his hope and quietly changes lives over decades")
```

```python
def predict_genre(svc_model,description):
  processed_des = preprocess_text(description)
  X_input = tfidf_vectorizer.transform([processed_des])
  predicted_genre_encoded = svc_model.predict(X_input)
  predicted_genre = encider.inverse_transform(predicted_genre_encoded)
  return predicted_genre[0]

predicted_genre = predict_genre(svc_model,new_des)
print(f"The predicted genre of the movie is: {predicted_genre}")
```
